## Vector Transposition

```python
## Transposing a regular 1-D array has no effect
x_trans = x.T

x_trans
```

```python
x_trans.shape
```

```python
## ...but it does we use nested "matrix-style" bracket
y = np.array([[25, 2, 5]])

y
```

```python
y.shape
```

```python
y.ndim
```

```python
## ...but can transpose a matrix with a dimension of length 1, which is mathmatically equivalent;

y_trans = y.T

y_trans
```

```python
y_trans.T
```

```python
y_trans.shape
```

## Norms and Unit Vectors

### L2 Norm

```python
x = np.array([25, 2, 5])

x
```

```python
(25**2 + 2**2 + 5**2) ** (1/2)
```

```python
np.linalg.norm(x)
```

### L1 Norm

```python
(25 + 2 + 5)
```

```python
np.abs(25) + np.abs(2) + np.abs(5)
```

### Squared L1 Norm

```python
(25**2 + 2**2 + 5**2)
```

```python
## we will cover tensor multiplication more soon but to prove point quickly
np.dot(x, x)
```

### Max Norm

```python
x
```

```python
np.max([np.abs(25), np.abs(2), np.abs(5)])
```
