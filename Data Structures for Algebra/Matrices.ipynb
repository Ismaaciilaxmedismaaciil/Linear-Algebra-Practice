# Matrix Tensors 

## Matrix (Rank 2 Tensor) In NumPy

```python
import numpy as np
```

```python
X = np.array([[1, 2], 
              [3, 4], 
              [5, 6]])

X
```

```python
X.shape
```

```python
X.ndim
```

```python
type(X)
```

```python
X.size
```

```python
X
```

```python
## Select left column of the matrix (Zero Indexed)
X[:, 0]
```

```python
## Select the middle row of matrix X
X[1, :]
```

```python
X
```

```python
## Another slicing-by-index; example
X[2:, :]
```

### Matrices in Pytorch

```python
import torch
```

```python
X_pt = torch.tensor([[10, 11], [12, 13], [14, 15]])

X_pt
```

```python
X_pt.shape
```

```python
X_pt[1:2, :]
```

### Matrices in TensorFlow

```python
import tensorflow as tf
```

```python
X_tf = tf.Variable([[10, 11], [12, 13], [14, 15]])

X_tf
```

```python
tf.rank(X_tf)
```

```python
tf.shape(X_tf)
```

```python
X_tf[1:2, :]
```
