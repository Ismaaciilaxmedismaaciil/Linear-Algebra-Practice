# Basic Tensor Arthmetic (The Hadamard Product)

## Basic Tensor Arthmetic

Adding or multiplying with scalar applies operation to all elements and tensor shape is retained.

```python
import numpy as np
```

```python
X = np.array([[25, 2], [5, 26], [3, 7]])

X
```

```python
X * 2
```

```python
X + 2
```

```python
X * 2 + 2
```

```python
import torch
```

```python
X_pt = torch.tensor([[25, 2], [5, 26], [3, 7]])

X_pt
```

```python
X_pt * 2 + 2  # Python operation are overloaded; could alternatively use torch.mul() or torch.add()
```

```python
torch.add(torch.mul(X_pt, 2), 2)
```

```python
import tensorflow as tf
```

```python
X_tf = tf.Variable([[25, 2], [5, 26], [3, 7]])

X_tf
```

```python
X_tf * 2 + 2  # Likewise overloaded; equavalent to tf.multiply() or tf.add()
```

```python
tf.add(tf.multiply(X_tf, 2), 2)
```

## The Hadamard Product

If two tensors have the same size, operations are often by defualt applied element-wise. This is not matrix multiplication, which we will cover latter The `Hadamard Product` or `Simply Element-wise Product`.

The mathmetical notation is A **.** X

```python
X
```

```python
A = X + 2

A
```

```python
A + X
```

```python
A * X
```

```python
A_pt = X_pt + 2

A_pt
```

```python
A_pt + X_pt
```

```python
A_pt * X_pt
```

```python
A_tf = X_tf + 2

A_tf
```

```python
A_tf + X_tf
```

```python
A_tf * X_tf
```
