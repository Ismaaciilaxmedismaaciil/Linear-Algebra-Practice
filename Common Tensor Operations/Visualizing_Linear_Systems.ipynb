# Visualizing Linear Systems

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Solving Linear Systems

### In Elimination Method:-

<img src="../Images/Elimination_method.png" alt="Sawirkayga" width="700">

```python
x = np.linspace(-10, 10, 1000)  #start, finish, n-points

y1 = 3 * x

y2 = 1 + (5*x) / 2
```

```python
fig, ax = plt.subplots()
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
ax.set_xlim([0, 3])
ax.set_ylim([0, 8])
ax.plot(x, y1, c='green')
ax.plot(x, y2, c='brown')
plt.axvline(x=2, color='purple', linestyle='--')
_ = plt.axhline(y=6, color='purple', linestyle='--')

plt.show()
```

### In Substitution Method:-

<img src="../Images/Substitution_method.png" alt="Sawirkayga" width="700">

```python
y1 = -5 + (2*x) / 3
y2 = (7 - 2 * x) / 5
```

```python
fig, ax = plt.subplots()
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')

ax.set_xlim([-2, 10])
ax.set_ylim([-6, 4])

## Adding x and y axes:
plt.axvline(x=0, color='lightgray')
plt.axhline(y=0, color='lightgray')

ax.plot(x, y1, c='green')
ax.plot(x, y2, c='brown')

plt.axvline(x=6, color='purple', linestyle='--')
plt.axhline(y=-1, color='purple', linestyle='--')

plt.show()
```
