# The Dot Product

Id we have two vectors (say, x and y) with the same legth n, we can calculate the dot ptoduct between them. This is annotated several different ways, including the following:

 - A) x **.** y
 - B) x(Transpose) * y 
 - C) (x, y) 

### Example

```python
import numpy as np
```

```python
x = np.array([25, 2, 5])

x
```

```python
y = np.array([0, 1, 2])

y
```

```python
25*0 + 2*1 + 5*2
```

```python
np.dot(x, y)  
```

```python
import torch
```

```python
x_pt = torch.tensor([25, 2, 5])


x_pt
```

```python
y_pt = torch.tensor([0, 1, 2])

y_pt
```

```python
np.dot(x_pt, y_pt)
```

```python
torch.dot(torch.tensor([25, 2, 5.]), torch.tensor([0, 1, 2.]))  
# Waxaa qasab ah marka dot isticmaleysid torch inaad arrayda haysatid ay float tahay.
```

```python
import tensorflow as tf
```

```python
x_tf = tf.Variable([25, 2, 5])

x_tf
```

```python
y_tf = tf.Variable([0, 1, 2])

y_tf
```

```python
tf.reduce_sum(tf.multiply(x_tf, y_tf))
```
