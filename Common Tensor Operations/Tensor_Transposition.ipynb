# Tensor Transpositions

### Tensor Transposition in NumPy

```python
import numpy as np
```

```python
X = np.array([[25, 2], [5, 26], [3, 7]])

X
```

```python
## Transpose
X.T
```

### Tensor Transposition In TensorFlow

```python
import tensorflow as tf
```

```python
X_tf = tf.Variable([[25, 2], [5, 26], [3, 7]])

X_tf
```

```python
## Transpose
tf.transpose(X_tf)
```

### Tensor Transposition In Pytorch

```python
import torch
```

```python
X_pt = torch.tensor([[25, 2], [5, 26], [3, 7]])

X_pt
```

```python
## Transpose
X_pt.T
```
