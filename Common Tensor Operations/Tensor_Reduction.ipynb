# Tensor Reduction

Calculating the sum accross all elements of a tensor is a common operations

```python
import numpy as np
```

```python
X = np.array([[25, 2], [5, 26], [3, 7]])

X
```

```python
X.sum()
```

```python
import torch
```

```python
X_pt = torch.tensor([[25, 2], [5, 26], [3, 7]])

X_pt
```

```python
torch.sum(X_pt)
```

```python
import tensorflow as tf
```

```python
X_tf = tf.Variable([[25, 2], [5, 26], [3, 7]])

X_tf
```

```python
tf.reduce_sum(X_tf)
```

Can also be dne along one specific axis alone, e.g:

```python
X.sum(axis=0)  #summing all rows
```

```python
X.sum(axis=1) #summing all columns
```

```python
torch.sum(X_pt, 0)
```

```python
torch.sum(X_pt, 1)
```

```python
tf.reduce_sum(X_tf, 1)
```

Many other operations can be applied with reduction along all or a selection of axes, e.g.:

- A) Maximum 
- B) Minimum
- C) Mean
- D) Product

They are fairly straight forward and used less often then summation, so you are welcome to look them up in library doc's if you are ever need them.
