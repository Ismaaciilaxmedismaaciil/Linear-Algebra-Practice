# Frobenius Norm

- Analogous to L2 norm of vector.
- Measures the size of matrix in terms of euclidean distance.
   - It's the sum of the magnitude of all the vectors in **X**

```python
import numpy as np
import tensorflow as tf
import torch
```

```python
X = np.array([[1, 2], [3, 4]])

X
```

```python
## Manually written
(1**2 + 2**2 + 3**2 + 4**2) ** (1/2)
```

```python
np.linalg.norm(X)
```

```python
X_pt = torch.tensor([[1, 2], [3, 4.]])  ## torch.norm requires float type

torch.norm(X_pt)
```

```python
X_tf = tf.Variable([[1, 2], [3, 4.]]) ## tf.norm also requires float dtype

tf.norm(X_tf)
```
