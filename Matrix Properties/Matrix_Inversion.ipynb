# Matrix Inversion

```python
import numpy as np
import tensorflow as tf
import torch
```

#### In NumPy

```python
## Matrix data / Input data
X = np.array([[4, 2], [-5, -3]])

X
```

```python
## Matrix Inverse data
Xinv = np.linalg.inv(X)

Xinv
```

```python
## y data / Outcames data
y = np.array([4, -7])

y
```

```python
## w (weights vector)
w = np.dot(Xinv, y)

w
```

```python
## Hada aan resultiga eegno y 
np.dot(X, w)
```

#### In Pytorch and TensorFlow

```python
torch.inverse(torch.tensor([[4, 2], [-5, -3.]])) ## float type
```

```python
tf.linalg.inv(tf.Variable([[4, 2], [-5, -3.]]))  # Also float type
```
