# Matrix-yada Orthogonal-ka ah

Matrix waa **orthogonal** marka tiirarkiisu yihiin **orthonormal**:

1. **Orthogonal:** laba tiir kasta oo kala duwan, dot product-kooduna waa $0$.
2. **Unit norm:** tiir kasta dhererkiisa waa $1$, markaas $\|v\| = \sqrt{v \cdot v} = 1$.

Si kooban: $Q^T Q = I$.

---

## Layliga 1 iyo 2: $I_3$ (qalin iyo warqad)

Tiirarka $I_3$ waa:

$$e_1 = (1, 0, 0),\quad e_2 = (0, 1, 0),\quad e_3 = (0, 0, 1)$$

**Dot products (1):**

- $e_1 \cdot e_2 = (1)(0) + (0)(1) + (0)(0) = 0$
- $e_1 \cdot e_3 = (1)(0) + (0)(0) + (0)(1) = 0$
- $e_2 \cdot e_3 = (0)(0) + (1)(0) + (0)(1) = 0$

Sidaas darteed, tiir kasta waa orthogonal kan kale.

**Norms (2):**

- $\|e_1\| = \sqrt{1^2 + 0^2 + 0^2} = 1$
- $\|e_2\| = \sqrt{0^2 + 1^2 + 0^2} = 1$
- $\|e_3\| = \sqrt{0^2 + 0^2 + 1^2} = 1$

Sidaas darteed, tiir kastaa wuxuu leeyahay unit norm.

---

## Layliga 3: NumPy ($I_3$)

```python
import numpy as np

I = np.eye(3)

# (1) Dot product-ka u dhexeeya tiirarka
for i in range(3):
    for j in range(i + 1, 3):
        print(f"col{i+1} · col{j+1} =", np.dot(I[:, i], I[:, j]))

# (2) Norm-ka tiir kasta
for i in range(3):
    print(f"||col{i+1}|| =", np.linalg.norm(I[:, i]))

# Hubin degdeg ah: Q^T Q = I
print(np.allclose(I.T @ I, np.eye(3)))
```

## Layliga 4: Matrix $K$

$$K = \begin{bmatrix} 2/3 & 1/3 & 2/3 \\ -2/3 & 2/3 & 1/3 \\ 1/3 & 2/3 & -2/3 \end{bmatrix}$$

Tiirarkiisu waa:

$$k_1 = \left(\tfrac{2}{3}, -\tfrac{2}{3}, \tfrac{1}{3}\right),\quad k_2 = \left(\tfrac{1}{3}, \tfrac{2}{3}, \tfrac{2}{3}\right),\quad k_3 = \left(\tfrac{2}{3}, \tfrac{1}{3}, -\tfrac{2}{3}\right)$$

### (1) Dot products (qalin iyo warqad)

- $k_1 \cdot k_2 = \tfrac{2}{9} - \tfrac{4}{9} + \tfrac{2}{9} = 0$
- $k_1 \cdot k_3 = \tfrac{4}{9} - \tfrac{2}{9} - \tfrac{2}{9} = 0$
- $k_2 \cdot k_3 = \tfrac{2}{9} + \tfrac{2}{9} - \tfrac{4}{9} = 0$

Dhammaan waa $0$, markaa tiirarku waa orthogonal.

### (2) Norms (qalin iyo warqad)

- $\|k_1\|^2 = \tfrac{4}{9} + \tfrac{4}{9} + \tfrac{1}{9} = 1$, markaa $\|k_1\| = 1$
- $\|k_2\|^2 = \tfrac{1}{9} + \tfrac{4}{9} + \tfrac{4}{9} = 1$, markaa $\|k_2\| = 1$
- $\|k_3\|^2 = \tfrac{4}{9} + \tfrac{1}{9} + \tfrac{4}{9} = 1$, markaa $\|k_3\| = 1$

Dhammaan waa $1$.

### (3) NumPy

```python
import numpy as np

K = np.array([[ 2/3, 1/3,  2/3],
              [-2/3, 2/3,  1/3],
              [ 1/3, 2/3, -2/3]])

# (1) Dot products
for i in range(3):
    for j in range(i + 1, 3):
        print(f"col{i+1} · col{j+1} =", np.round(np.dot(K[:, i], K[:, j]), 10))

# (2) Norms
for i in range(3):
    print(f"||col{i+1}|| =", np.linalg.norm(K[:, i]))

# Hubin guud: K^T K = I
print(np.allclose(K.T @ K, np.eye(3)))
```

## Gabagabo

Matrix $K$ **waa orthogonal**, sababtoo ah tiirarkiisu way is-orthogonal yihiin, mid kastana wuxuu leeyahay unit norm, sidaa darteed $K^T K = I_3$. Tani waxay kaloo la macno tahay $K^{-1} = K^T$.
