# Matrix Multiplication

## Matrix by Vector

```python
import numpy as np 
import tensorflow as tf 
import torch 
```

```python
## Creating a matrix called 'A' using numpy
A = np.array([[3, 4], [5, 6], [7, 8]])

A
```

```python
## Creating a vector called 'b'
b = np.array([1, 2])

b
```

```python
## Multiplication matrix(A) by vector(b), Using dot product
np.dot(A, b)
```

```python
# In Pytorch
## Creating a matrix called 'A_pt', Using pytorch
A_pt = torch.tensor([[3, 4], [5, 6], [7, 8]])

A_pt
```

```python
## Creating a vector called 'b_pt', Using pytoch
b_pt = torch.tensor([1, 2])

b_pt
```

```python
## Mutiplying a matric and vector, Using torch.matmul()
torch.matmul(A_pt, b_pt)
```

```python
# In TensorFlow
## Creating a matrix called 'A_tf', Using tensoflow
A_tf = tf.Variable([[3, 4], [5, 6], [7, 8]])

A_tf
```

```python
## Creating a vector called 'b_tf', Using tensorflow
b_tf = tf.Variable([1, 2])

b_tf
```

```python
## Multiplying a matrix and vector, Using tf.linalg.matvec()
tf.linalg.matvec(A_tf, b_tf)
```

## Matrix by Matrix multiplication

```python
A
```

```python
## Creating matrix two called 'B', Using Numpy
B = np.array([[1, 9], [2, 0]])
B
```

```python
## Multiplying the two matrices, Using np.dot()
AB = np.dot(A, B)

AB
```

Note that matrix multiplication is not `Commutative` '(i.e., AB != BA)', So uncommenting the following line will though a size missmatch.

```python
# np.dot(B, A)  ## Size Missmatch Error
```

```python
B_pt = torch.from_numpy(B)  ## Much cleaner then TF conversion

B_pt
```

```python
## Another neat way to create the same tensor with transposition.
B_pt = torch.tensor([[1, 2], [9, 0]]).T

B_pt
```

```python
## Multiplying the two matrices, Using torch.matmul()
torch.matmul(A_pt, B_pt)
```

```python
B_tf = tf.convert_to_tensor(B, dtype=tf.float32)

B_tf
```

```python
## Same datatype ka wada dhig ka dib isku dhufo
A_tf = tf.cast(A_tf, tf.float32)
B_tf = tf.cast(B_tf, tf.float32)

tf.matmul(A_tf, B_tf)
```
