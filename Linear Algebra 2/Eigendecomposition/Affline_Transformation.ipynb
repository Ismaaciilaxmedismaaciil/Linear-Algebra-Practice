# Affline Transformation via Matrix Application

Let's we have a vector `v`

```python
import numpy as np
```

```python
v = np.array([3, 1])
v
```

Let's plot `v` using my `plot_vector()` function

```python
import matplotlib.pyplot as plt
```

```python
def plot_vector(vectors, colors):
    plt.figure()
    plt.axvline(x=0, color="lightgray")
    plt.axhline(y=0, color="lightgray")

    for vector, color in zip(vectors, colors):
        plt.quiver(
            [0], [0], [vector[0]], [vector[1]],
            angles="xy",
            scale_units="xy",
            scale=1,
            color=color,
        )

    plt.xlim(-1, 5)
    plt.ylim(-1, 5)
    plt.gca().set_aspect("equal")

plot_vector([v], ["lightblue"])
```

"Applying" a matrix to a vector (i.e., performing matrix-vector transformation) can linearly transform the vector, e.g., rotate it or rescale it.


The identity matrix, introdiced earlier, is the exception that provides the rule: Applying an identity  matrix does not transform  the vector.

```python
I = np.array([[1, 0], [0, 1]])

I
```

```python
Iv = np.dot(I, v)

Iv
```

```python
v == Iv
```

```python
plot_vector([Iv], ['blue'])
plt.xlim(-1, 5)
_ = plt.ylim(-1, 5)
```
