# Applying Matrices

## Matrix-Application Exercises

```python
import numpy as np
```

```python
# Vector u1
u1 = np.array([2, 5, -3])
u1
```

```python
# Vector u2
u2 = np.array([0, -4, 6])
u2
```

```python
# Matrix B
B = np.array([[2, 0, -1], [-2, 3, 1], [0, 4, -1]])
B
```

#### Adigoo Isticmaalaya vector u1, vector u2, iyo matrix B. 

**Xisaabi su.alaahan soo socda:-**

 1) Apply the Identity matrix `I3` to the vector `u1`.
2) Apply the matrix `B` to the vector `u1`.
3) Concatenate vector `u1` with vector `u2` to form a matrix `U`, then apply the matrix `B` to the matrix `U`.

```python
## Jawabta su.aasha 1aad
I = np.array([[1, 0, 0], [0, 1, 0], [0, 0, 1]])
I

Iu1 = np.dot(u1, I)

Iu1
```

```python
## Jawaabta su,aasha 2aad
Bu1 = np.dot(B, u1)

Bu1
```

```python
## Jawaabta su,aasha 3aad

## Marka hore isku xir labada vector oo kala ah u1, u2

result = np.vstack((u1, u2))

U = result.T

U
```

```python
## Ka dib soo saar Bu2

Bu2 = np.dot(B, u2)

Bu2
```

```python
## Ka dib isku xir Bu1 iyo Bu2, si uu usoo baxo hal matrix BU
BU_ver_Stack = np.vstack((Bu1, Bu2))

BU = BU_ver_Stack.T

BU
```

```python
## Another way
np.dot(B, U)
```
